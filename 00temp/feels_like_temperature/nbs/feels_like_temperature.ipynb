# Feels Like Temperature 验证

官方测试数据为**投影坐标**网格；KGO 也是在该投影输入上得到的结果。因此**优先用投影坐标验证**修改后方法是否与原方法 / KGO 一致。

同时需要保证算法能接受**经纬输入**：将原测试数据投影转经纬并重网格后，再分别作为原方法（Iris Cube）与修改后方法（meb 六维）的输入，对比两边结果，确认方法修改没有引入错误。

对照三条结果：

1. **修改后方法**（meb 六维 / `calculate_feels_like_temperature`）
2. **原方法**（Iris Cube / IMPROVER 同名函数）
3. **KGO**（官方样例；仅方案一与投影输入直接比数值）

数据来自 `test_data/feels_like_temp_data/`。请先运行：

```text
python feels_like_temperature/cli/preprocess_test_data.py
```

生成 `cli_input/`（方案一：投影维重命名 meb）与 `latlon/` + `latlon/cli_input/`（方案二：经纬 Cube / 同源 meb）。


## 1. 环境准备


````python
%matplotlib inline

import sys
import warnings
from pathlib import Path

import iris
import matplotlib.pyplot as plt
import meteva_base as meb
import numpy as np
import xarray as xr

warnings.filterwarnings("ignore", category=FutureWarning)
warnings.filterwarnings("ignore", message="pkg_resources is deprecated")

plt.rcParams["font.sans-serif"] = ["SimHei", "Microsoft YaHei", "SimSun", "DejaVu Sans"]
plt.rcParams["axes.unicode_minus"] = False

PACKAGE_ROOT = Path.cwd().resolve().parents[0]
PROJECT_ROOT = PACKAGE_ROOT.parent
IMPROVER_ROOT = PROJECT_ROOT / "improver-1.18.7"
DATA_DIR = PACKAGE_ROOT / "test_data" / "feels_like_temp_data"
CLI_INPUT_DIR = DATA_DIR / "cli_input"
CLI_OUTPUT_DIR = DATA_DIR / "cli_output"
LATLON_DIR = DATA_DIR / "latlon"
LATLON_MEB_DIR = LATLON_DIR / "cli_input"

for path in (str(PROJECT_ROOT), str(IMPROVER_ROOT)):
    if path not in sys.path:
        sys.path.insert(0, path)

from improver.temperature.feels_like_temperature import (  # noqa: E402
    calculate_feels_like_temperature as orig_calculate_feels_like_temperature,
)
from feels_like_temperature.src.feels_like_temperature import (  # noqa: E402
    calculate_feels_like_temperature,
)
from feels_like_temperature.cli.der_feel_like_temp import process as cli_flt_process  # noqa: E402

INPUT_FILES = {
    "temperature": "20181121T1200Z-PT0012H00M-temperature_at_screen_level.nc",
    "wind_speed": "20181121T1200Z-PT0012H00M-wind_speed_at_10m.nc",
    "relative_humidity": "20181121T1200Z-PT0012H00M-relative_humidity_at_screen_level.nc",
    "pressure": "20181121T1200Z-PT0012H00M-pressure_at_mean_sea_level.nc",
}

print(f"包目录: {PACKAGE_ROOT}")
print(f"数据目录: {DATA_DIR}")
print(f"IMPROVER: {IMPROVER_ROOT}")
````

## 2. 公共工具


````python
def require_files(paths, label: str) -> None:
    missing = [str(p) for p in paths if not Path(p).exists()]
    if missing:
        raise FileNotFoundError(
            f"缺少{label}（共 {len(missing)} 个）。"
            "请先运行: python feels_like_temperature/cli/preprocess_test_data.py\n"
            + "\n".join(missing[:8])
        )


def load_meb(path: Path) -> xr.DataArray:
    return meb.checkout_griddata(
        meb.read_griddata_from_nc(str(path)),
        valid_val=(-np.inf, np.inf, np.nan),
    )


def load_cube(path: Path):
    return iris.load_cube(str(path))


def spatial_values(data) -> np.ndarray:
    if isinstance(data, xr.DataArray):
        arr = np.asarray(data.values, dtype=np.float64)
    elif hasattr(data, "data"):
        raw = data.data
        arr = (
            np.ma.filled(np.ma.asarray(raw, dtype=np.float64), np.nan)
            if np.ma.isMaskedArray(raw)
            else np.asarray(raw, dtype=np.float64)
        )
    else:
        arr = np.asarray(data, dtype=np.float64)
    return np.squeeze(arr)


def compare_fields(current, reference, tag: str, *, rtol: float = 1e-5, atol: float = 1e-4) -> dict:
    c = spatial_values(current)
    r = spatial_values(reference)
    if c.shape != r.shape:
        raise ValueError(f"形状不一致: current={c.shape}, reference={r.shape}, tag={tag}")

    both_nan = np.isnan(c) & np.isnan(r)
    finite = np.isfinite(c) & np.isfinite(r)
    only_c = np.isfinite(c) & np.isnan(r)
    only_r = np.isnan(c) & np.isfinite(r)
    diff = np.zeros_like(c, dtype=np.float64)
    diff[finite] = c[finite] - r[finite]

    stats = {
        "allclose": bool(np.allclose(c, r, rtol=rtol, atol=atol, equal_nan=True)),
        "mean_abs": float(np.mean(np.abs(diff[finite]))) if np.any(finite) else 0.0,
        "max_abs": float(np.max(np.abs(diff[finite]))) if np.any(finite) else 0.0,
        "rmse": float(np.sqrt(np.mean(diff[finite] ** 2))) if np.any(finite) else 0.0,
        "nan_mismatch": int(only_c.sum() + only_r.sum()),
        "both_nan": int(both_nan.sum()),
    }
    print(
        f"[{tag}] allclose={stats['allclose']}, "
        f"mean_abs={stats['mean_abs']:.8f}, max_abs={stats['max_abs']:.8f}, "
        f"rmse={stats['rmse']:.8f}, nan_mismatch={stats['nan_mismatch']}"
    )
    return stats


def plot_inputs(fields: dict, title: str) -> None:
    fig, axes = plt.subplots(2, 2, figsize=(10, 8), constrained_layout=True)
    for ax, (name, data) in zip(axes.ravel(), fields.items()):
        im = ax.imshow(spatial_values(data), origin="lower", cmap="viridis", aspect="auto")
        ax.set_title(name)
        fig.colorbar(im, ax=ax, fraction=0.046, pad=0.04)
    fig.suptitle(title)
    plt.show()


def plot_triplet(modified, reference, extra, titles, title: str, *, extra_is_diff: bool = False) -> None:
    arrays = [spatial_values(modified), spatial_values(reference), spatial_values(extra)]
    fig, axes = plt.subplots(1, 3, figsize=(15, 4.8), constrained_layout=True)
    clim_src = arrays[:2] if extra_is_diff else arrays
    vmin = min(float(np.nanmin(a)) for a in clim_src)
    vmax = max(float(np.nanmax(a)) for a in clim_src)
    for i, (ax, data, name) in enumerate(zip(axes, arrays, titles)):
        cmap = "coolwarm" if extra_is_diff and i == 2 else "turbo"
        kw = {"origin": "lower", "cmap": cmap, "aspect": "auto"}
        if not (extra_is_diff and i == 2):
            kw["vmin"] = vmin
            kw["vmax"] = vmax
        im = ax.imshow(data, **kw)
        ax.set_title(name)
        fig.colorbar(im, ax=ax, fraction=0.046, pad=0.04)
    fig.suptitle(title)
    plt.show()


print("公共工具已加载。")
````

## 3. 方案一：投影对齐输入（`cli_input/`）

投影维仅重命名为 `lat`/`lon`（数值仍为米制）。修改后读 meb；原方法读官方投影 Cube；KGO 直接比数值。


````python
meb_paths = {key: CLI_INPUT_DIR / name for key, name in INPUT_FILES.items()}
cube_paths = {key: DATA_DIR / name for key, name in INPUT_FILES.items()}
kgo_path = DATA_DIR / "kgo.nc"
original_path = DATA_DIR / "original_feels_like_temp_result.nc"
original_cli_path = DATA_DIR / "original_cli_result.nc"

require_files(list(meb_paths.values()), "方案一 meb 输入")
require_files(list(cube_paths.values()) + [kgo_path, original_path, original_cli_path], "方案一官方对照")

t_meb = load_meb(meb_paths["temperature"])
w_meb = load_meb(meb_paths["wind_speed"])
rh_meb = load_meb(meb_paths["relative_humidity"])
p_meb = load_meb(meb_paths["pressure"])

t_cube = load_cube(cube_paths["temperature"])
w_cube = load_cube(cube_paths["wind_speed"])
rh_cube = load_cube(cube_paths["relative_humidity"])
p_cube = load_cube(cube_paths["pressure"])
kgo = load_cube(kgo_path)
original_saved = load_cube(original_path)

print("方案一 meb:", t_meb.dims, t_meb.shape)
print("方案一 Cube:", t_cube.shape, t_cube.coord_system())
````

````python
# 适配修改后方法调用
modified_s1 = calculate_feels_like_temperature(t_meb, w_meb, rh_meb, p_meb)
# 原方法调用
original_s1 = orig_calculate_feels_like_temperature(t_cube, w_cube, rh_cube, p_cube)

print("=== 方案一：体感温度（投影对齐）===")
compare_fields(modified_s1, original_s1, "方案一 修改后-原方法")
compare_fields(modified_s1, kgo, "方案一 修改后-KGO")
compare_fields(original_s1, kgo, "方案一 原方法-KGO")
compare_fields(original_s1, original_saved, "方案一 原方法-预存原结果")

plot_inputs(
    {"气温": t_meb, "风速": w_meb, "相对湿度": rh_meb, "气压": p_meb},
    "方案一输入场（投影维重命名 meb）",
)
plot_triplet(
    modified_s1,
    original_s1,
    kgo,
    ("修改后", "原方法", "KGO"),
    "方案一：结果对比（投影对齐）",
)
````

## 4. 方案二：经纬重网格输入（`latlon/` + `latlon/cli_input/`）

- **原方法**：读 `latlon/*.nc`（经纬 Cube）
- **修改后**：读 `latlon/cli_input/*.nc`（同源数值的 meb）

两边仅格式不同。主判据是修改后 vs 原方法。


````python
ll_cube_paths = {key: LATLON_DIR / name for key, name in INPUT_FILES.items()}
ll_meb_paths = {key: LATLON_MEB_DIR / name for key, name in INPUT_FILES.items()}

require_files(list(ll_cube_paths.values()), "方案二经纬 Cube")
require_files(list(ll_meb_paths.values()), "方案二经纬 meb")

t_ll_meb = load_meb(ll_meb_paths["temperature"])
w_ll_meb = load_meb(ll_meb_paths["wind_speed"])
rh_ll_meb = load_meb(ll_meb_paths["relative_humidity"])
p_ll_meb = load_meb(ll_meb_paths["pressure"])

t_ll_cube = load_cube(ll_cube_paths["temperature"])
w_ll_cube = load_cube(ll_cube_paths["wind_speed"])
rh_ll_cube = load_cube(ll_cube_paths["relative_humidity"])
p_ll_cube = load_cube(ll_cube_paths["pressure"])

print("方案二 meb:", t_ll_meb.dims, t_ll_meb.shape)
print("lat 范围:", float(t_ll_meb.lat.min()), float(t_ll_meb.lat.max()))
print("lon 范围:", float(t_ll_meb.lon.min()), float(t_ll_meb.lon.max()))
print("方案二 Cube:", t_ll_cube.shape, [c.name() for c in t_ll_cube.dim_coords])
````

````python
# 适配修改后方法调用
modified_s2 = calculate_feels_like_temperature(t_ll_meb, w_ll_meb, rh_ll_meb, p_ll_meb)
# 原方法调用
original_s2 = orig_calculate_feels_like_temperature(t_ll_cube, w_ll_cube, rh_ll_cube, p_ll_cube)

print("=== 方案二：体感温度（经纬网格）===")
compare_fields(modified_s2, original_s2, "方案二 修改后-原方法")
compare_fields(t_ll_meb, t_ll_cube, "方案二 输入气温 meb-Cube")

plot_inputs(
    {"气温": t_ll_meb, "风速": w_ll_meb, "相对湿度": rh_ll_meb, "气压": p_ll_meb},
    "方案二输入场（经纬 meb）",
)
diff_s2 = spatial_values(modified_s2) - spatial_values(original_s2)
plot_triplet(
    modified_s2,
    original_s2,
    diff_s2,
    ("修改后", "原方法", "修改后-原方法"),
    "方案二：结果对比（经纬网格）",
    extra_is_diff=True,
)
````

## 5. CLI 应用示例与结果对比

CLI 从 `cli_input/` 读取方案一预处理输入，调用 `der_feel_like_temp.process`，并与原 CLI、KGO 对比。


````python
cli_result = cli_flt_process(
    str(meb_paths["temperature"]),
    str(meb_paths["wind_speed"]),
    str(meb_paths["relative_humidity"]),
    str(meb_paths["pressure"]),
    output_path=None,
)
original_cli = load_cube(original_cli_path)

print("=== CLI：体感温度 ===")
compare_fields(cli_result, modified_s1, "CLI-修改后插件")
compare_fields(cli_result, original_cli, "CLI-原算法CLI")
compare_fields(cli_result, kgo, "CLI-KGO")
compare_fields(original_cli, kgo, "原算法CLI-KGO")

plot_triplet(
    cli_result,
    original_cli,
    kgo,
    ("修改后 CLI", "原算法 CLI", "KGO"),
    "CLI 结果对比",
)
````

## 6. 小结

| 方案 | 修改后输入 | 原方法输入 | 主判据 |
| --- | --- | --- | --- |
| 方案一 | `cli_input/` meb | 官方投影 Cube | 修改后 vs 原方法；KGO 作回归对照 |
| 方案二 | `latlon/cli_input/` meb | `latlon/` Cube | **修改后 vs 原方法**（同源经纬） |

方案一用于确认算法与原版/KGO 的数值一致；方案二用于确认真经纬 Cube 与同源 meb 可得到一致结果。
