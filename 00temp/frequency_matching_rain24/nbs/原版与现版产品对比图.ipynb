# 原版与现版频率匹配产品对比图

加载已生成的对比图（左：原版，中：现版，右：现版−原版），路径：
`resource/output/pngs/original_vs_aligned/{起报}/{起报}.{时效}.png`。

差异说明见 `docs/原版original与现版aligned频率匹配差异与应用场景.md`。

```python
from pathlib import Path

import matplotlib.pyplot as plt
from IPython.display import display
from PIL import Image

ROOT = Path("..").resolve()
PNG_ROOT = ROOT / "resource" / "output" / "pngs" / "original_vs_aligned"

# 短时效 / 中时效各选一张，便于对照
SHOW = [
    ("2026081300", 36),
    ("2026081300", 72),
    ("2026081300", 120),
    ("2026081312", 36),
]

paths = []
for rpt, dtime in SHOW:
    p = PNG_ROOT / rpt / f"{rpt}.{dtime:03d}.png"
    if p.is_file():
        paths.append(p)
    else:
        print("missing:", p)

print("found", len(paths), "figures under", PNG_ROOT)
for p in paths:
    print("-", p.relative_to(ROOT))
```

```python
for p in paths:
    img = Image.open(p)
    fig, ax = plt.subplots(figsize=(12, 3.6))
    ax.imshow(img)
    ax.set_axis_off()
    ax.set_title(p.name, fontsize=12)
    display(fig)
    plt.close(fig)
```
