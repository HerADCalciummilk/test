# orographic_enhancement 模块演示与验证

本 Notebook 做两件事：

1. **演示怎么用**：读取已预处理的测试数据，调用两个算法，看输入/输出长什么样。输入适配修改后的方法简称为迁移方法。
2. **验证算得对不对**：把迁移后的结果，和原 IMPROVER 方法、官方样例（KGO）做数值对比。

两个算法分别是：

- `MetaOrographicEnhancement`：根据边界层温湿压和风，结合地形，**算出**地形对降水的增强项（单位 m s-1，图里会换成 mm hr-1 方便看）。
- `ApplyOrographicEnhancement`：已经有增强项之后，把它 **加到**（或从中减去）降水场上。叠加本身不读格距，纯经纬网格也能跑。

官方样例是**投影坐标**网格，KGO 也是在投影输入上算出来的，所以**先看投影坐标输入**。
后面的经纬坐标输入把同一套数据重网格到经纬。注意：原 IMPROVER 的 `OrographicEnhancement` 会把水平坐标换算成米再算梯度，**不能**直接吃纯经纬；经纬路径里计算增强项只跑迁移方法，并用重网格后的官方结果对照。`ApplyOrographicEnhancement` 原方法在经纬上可以现场对照。

请先在仓库根目录运行一次预处理（只需做一次，会写出 `cli_input/` 和 `latlon/`）：

```text
python orographic_precipitation_downscaling/cli/preprocess_test_data.py
```

## 2. 读数、对比、画图用的小工具

后面会反复读文件、把场压成二维、打印误差、画并排图。先写成函数，正文里只保留“读哪些文件、调用哪个算法”。

增强项官方单位是 m s-1，量级大约 1e-6；图上乘 `3600*1000` 换成 mm hr-1，并用离散降水色标。
Cube 与 meb 的南北方向可能相反，对比和画图都会先按坐标把轴对齐。


````python
def require_files(paths, label: str) -> None:
    """缺文件时给出明确提示，避免后面读盘才报一长串错误。"""
    missing = [str(p) for p in paths if not Path(p).exists()]
    if missing:
        raise FileNotFoundError(
            f"缺少{label}（共 {len(missing)} 个）。"
            "请先运行: python orographic_precipitation_downscaling/cli/preprocess_test_data.py\n"
            + "\n".join(missing[:8])
        )


def to_2d(data) -> np.ndarray:
    """把六维场或 Cube 压成一张二维图，方便比数值、画图。"""
    if not isinstance(data, xr.DataArray) and hasattr(data, "data"):
        raw = data.data
        arr = (
            np.ma.filled(np.ma.asarray(raw, dtype=np.float64), np.nan)
            if np.ma.isMaskedArray(raw)
            else np.asarray(raw, dtype=np.float64)
        )
        return np.squeeze(arr)
    arr = data
    for dim in list(arr.dims):
        if dim not in (
            "lat",
            "lon",
            "latitude",
            "longitude",
            "projection_y_coordinate",
            "projection_x_coordinate",
        ):
            arr = arr.isel({dim: 0}, drop=True)
    return np.squeeze(np.asarray(arr.values, dtype=np.float64))


def compare_stats(current, reference, tag: str):
    """打印两个场的差异。三个数都接近 0，说明两边几乎一样。

    Cube 与 meb 的 y 轴方向可能相反，先按坐标对齐再减，避免把「南北颠倒」当成算法误差。
    """
    left_da = align_spatial_to_reference(to_plot_2d(current), to_plot_2d(reference))
    right_da = ensure_2d(to_plot_2d(reference))
    left = np.asarray(left_da.values, dtype=np.float64)
    right = np.asarray(right_da.values, dtype=np.float64)
    if left.shape != right.shape:
        raise ValueError(f"形状不一致: {left.shape} vs {right.shape}（{tag}）")
    diff = left - right
    print(
        f"[{tag}] "
        f"平均绝对误差={np.mean(np.abs(diff)):.8f}, "
        f"最大绝对误差={np.max(np.abs(diff)):.8f}, "
        f"均方根误差={np.sqrt(np.mean(diff ** 2)):.8f}"
    )
    return left, right, diff


# 离散分级降水色标（单位 mm hr-1）；>50 用深蓝色
PRECIP_CLASS_BOUNDS = (0.0, 0.1, 2.5, 5.0, 10.0, 25.0, 50.0)
PRECIP_CLASS_RGB = (
    (255, 255, 255),
    (210, 245, 160),
    (150, 220, 95),
    (70, 185, 65),
    (35, 155, 40),
    (96, 184, 255),
)
precip_cmap = ListedColormap(np.array(PRECIP_CLASS_RGB, dtype=float) / 255.0)
precip_cmap.set_over((0.0, 0.0, 1.0))
main_norm = BoundaryNorm(PRECIP_CLASS_BOUNDS, precip_cmap.N, clip=False)
PRECIP_CBAR_TICKS = (0.1, 2.5, 5.0, 10.0, 25.0, 50.0)
MPS_TO_MMHR = 3600.0 * 1000.0  # m s-1 → mm hr-1


def _format_precip_cbar_tick(value: float) -> str:
    if value < 1.0 or not float(value).is_integer():
        return f"{value:.1f}"
    return str(int(value))


def add_precip_colorbar(fig, axes, mesh=None, *, label: str = "mm hr-1"):
    """为离散分级降水图添加统一色标。"""
    cbar = fig.colorbar(
        mesh if mesh is not None else plt.cm.ScalarMappable(norm=main_norm, cmap=precip_cmap),
        ax=axes,
        extend="max",
        boundaries=list(PRECIP_CLASS_BOUNDS),
        ticks=list(PRECIP_CBAR_TICKS),
        shrink=0.8,
    )
    cbar.set_ticklabels([_format_precip_cbar_tick(t) for t in PRECIP_CBAR_TICKS])
    cbar.set_label(label)
    return cbar


def ensure_2d(data: xr.DataArray) -> xr.DataArray:
    result = data.squeeze(drop=True)
    if result.ndim != 2:
        raise ValueError(f"Expected a 2D field, got dims={result.dims}")
    return result


def cube_to_spatial_da(cube) -> xr.DataArray:
    """把 Iris Cube 转成带空间坐标的 DataArray，避免 from_iris 依赖 dask 版本。"""
    y_coord = cube.coord(axis="y")
    x_coord = cube.coord(axis="x")
    raw = cube.data
    arr = (
        np.ma.filled(np.ma.asarray(raw, dtype=np.float64), np.nan)
        if np.ma.isMaskedArray(raw)
        else np.asarray(raw, dtype=np.float64)
    )
    arr = np.squeeze(arr)
    return xr.DataArray(
        arr,
        dims=(y_coord.name(), x_coord.name()),
        coords={
            y_coord.name(): np.asarray(y_coord.points),
            x_coord.name(): np.asarray(x_coord.points),
        },
        name=cube.name() or "field",
        attrs={"units": str(cube.units)},
    )


def to_plot_2d(data) -> xr.DataArray:
    """压到二维 DataArray；Iris Cube 先转 xarray。"""
    if not isinstance(data, xr.DataArray):
        data = cube_to_spatial_da(data)
    candidate_pairs = [
        ("projection_y_coordinate", "projection_x_coordinate"),
        ("lat", "lon"),
        ("latitude", "longitude"),
    ]
    dims_set = set(data.dims)
    selected_pair = None
    for y_name, x_name in candidate_pairs:
        if y_name in dims_set and x_name in dims_set:
            selected_pair = (y_name, x_name)
            break
    if selected_pair is None:
        selected_pair = (data.dims[-2], data.dims[-1])
    indexers = {dim: 0 for dim in data.dims if dim not in set(selected_pair)}
    reduced = data.isel(indexers, drop=True) if indexers else data
    return ensure_2d(reduced)


def align_spatial_to_reference(field: xr.DataArray, reference: xr.DataArray) -> xr.DataArray:
    """把场的空间轴翻转到与参考场一致，便于按格点比较和画在同一套轴上。"""
    field2d = ensure_2d(to_plot_2d(field)).copy()
    ref2d = ensure_2d(to_plot_2d(reference)).copy()
    y_name, x_name = field2d.dims
    ry, rx = ref2d.dims
    if field2d.shape != ref2d.shape:
        raise ValueError(f"空间场形状不一致: field={field2d.shape}, reference={ref2d.shape}")
    y_f = np.asarray(field2d.coords[y_name].values, dtype=np.float64)
    y_r = np.asarray(ref2d.coords[ry].values, dtype=np.float64)
    x_f = np.asarray(field2d.coords[x_name].values, dtype=np.float64)
    x_r = np.asarray(ref2d.coords[rx].values, dtype=np.float64)
    aligned = field2d
    if np.allclose(y_f, y_r, rtol=0, atol=1.0):
        pass
    elif np.allclose(y_f[::-1], y_r, rtol=0, atol=1.0):
        aligned = aligned.isel({y_name: slice(None, None, -1)}).assign_coords({y_name: y_r})
    else:
        raise ValueError("Y 坐标无法与参考场对齐")
    if np.allclose(x_f, x_r, rtol=0, atol=1.0):
        pass
    elif np.allclose(x_f[::-1], x_r, rtol=0, atol=1.0):
        aligned = aligned.isel({x_name: slice(None, None, -1)}).assign_coords({x_name: x_r})
    else:
        raise ValueError("X 坐标无法与参考场对齐")
    return aligned.rename({y_name: ry, x_name: rx})


def orient_for_plot(data_2d: xr.DataArray) -> xr.DataArray:
    result = ensure_2d(data_2d).copy()
    y_name, x_name = result.dims
    y_vals = result.coords[y_name].values
    x_vals = result.coords[x_name].values
    if y_vals.size > 1 and y_vals[0] > y_vals[-1]:
        result = result.isel({y_name: slice(None, None, -1)})
    if x_vals.size > 1 and x_vals[0] > x_vals[-1]:
        result = result.isel({x_name: slice(None, None, -1)})
    return result


def plot_field(
    ax,
    data,
    title: str,
    *,
    reference=None,
    cmap=precip_cmap,
    norm=main_norm,
    cbar_label=None,
):
    """画一张二维场。默认用降水分级色标；连续场传 ``norm=None``，需要时用 ``cbar_label`` 加色标。"""
    plot_data = to_plot_2d(data)
    if reference is not None:
        plot_data = align_spatial_to_reference(plot_data, reference)
    else:
        plot_data = orient_for_plot(plot_data)
    y_name, x_name = plot_data.dims
    x_vals = plot_data.coords[x_name].values
    y_vals = plot_data.coords[y_name].values
    mesh_kw = {"shading": "auto", "cmap": cmap}
    if norm is not None:
        mesh_kw["norm"] = norm
    mesh = ax.pcolormesh(x_vals, y_vals, plot_data.values, **mesh_kw)
    # ylim 取 (min, max)，图面始终南在下、北在上；y 递减时 orient_for_plot 已先把数组翻成升序
    ax.set_xlim(float(np.nanmin(x_vals)), float(np.nanmax(x_vals)))
    ax.set_ylim(float(np.nanmin(y_vals)), float(np.nanmax(y_vals)))
    ax.set_title(title, fontsize=11)
    ax.set_xlabel(x_name)
    ax.set_ylabel(y_name)
    ax.set_aspect("equal")
    if cbar_label:
        ax.tick_params(labelsize=8)
        cbar = plt.colorbar(mesh, ax=ax, shrink=0.85, pad=0.02)
        cbar.set_label(cbar_label, fontsize=9)
    return mesh


def to_mmhr(data) -> xr.DataArray:
    result = ensure_2d(to_plot_2d(data)).copy()
    result.values = np.asarray(result.values, dtype=np.float64) * MPS_TO_MMHR
    result.attrs["units"] = "mm hr-1"
    return result


def plot_precip_triplet(modified, kgo, original, *, title: str):
    fig, axes = plt.subplots(1, 3, figsize=(22, 6), constrained_layout=True)
    fig.suptitle(title, fontsize=14, y=1.02)
    plot_ref = to_plot_2d(kgo)
    plot_field(axes[0], to_mmhr(modified), "修改后算法", reference=plot_ref)
    plot_field(axes[1], to_mmhr(original), "原方法结果", reference=plot_ref)
    mesh = plot_field(axes[2], to_mmhr(kgo), "KGO 官方预期")
    add_precip_colorbar(fig, axes, mesh=mesh)
    plt.show()


def select_level_for_plot(data: xr.DataArray, level: float | None = 1000.0) -> xr.DataArray:
    """从多层场中选取指定高度层（米）用于绘图，默认与 ``boundary_height=1000`` 对齐。"""
    if level is None or "level" not in data.dims:
        return data
    level_values = np.asarray(data.coords["level"].values, dtype=np.float64)
    if level_values.size == 1:
        return data.isel(level=0, drop=True)
    idx = int(np.argmin(np.abs(level_values - level)))
    return data.isel(level=idx, drop=True)


print("工具函数已就绪。")
````

## 3. 投影坐标输入

这里读的是 `cli_input/`：投影坐标只是把维名改成了 `lat`/`lon`，**数值仍是米制投影**。
迁移方法用 `meb.read_griddata_from_nc` 读六维网格；KGO 和原方法结果是官方 Iris 格式，用 `iris.load_cube` 读。


### 3.1 读取预处理结果

气象场带 5 层高度（100–1500 m）；地形是 1 km 高分辨率网格，算法内部会把气象场重采样到地形网格。


````python
met_files = [
    CLI_INPUT_DIR / "temperature.nc",
    CLI_INPUT_DIR / "humidity.nc",
    CLI_INPUT_DIR / "pressure.nc",
    CLI_INPUT_DIR / "wind_speed.nc",
    CLI_INPUT_DIR / "wind_direction.nc",
    CLI_INPUT_DIR / "orography_uk-standard_1km.nc",
]
require_files(
    met_files + [DATA_DIR / "kgo_hi_res.nc", DATA_DIR / "original_algorithm_result.nc"],
    "投影坐标输入",
)

# 迁移方法只读预处理后的 meb；KGO / 原方法结果仍用官方 Cube
temperature_meb = meb.read_griddata_from_nc(str(CLI_INPUT_DIR / "temperature.nc"))
humidity_meb = meb.read_griddata_from_nc(str(CLI_INPUT_DIR / "humidity.nc"))
pressure_meb = meb.read_griddata_from_nc(str(CLI_INPUT_DIR / "pressure.nc"))
wind_speed_meb = meb.read_griddata_from_nc(str(CLI_INPUT_DIR / "wind_speed.nc"))
wind_direction_meb = meb.read_griddata_from_nc(str(CLI_INPUT_DIR / "wind_direction.nc"))
topography_meb = meb.read_griddata_from_nc(str(CLI_INPUT_DIR / "orography_uk-standard_1km.nc"))

kgo_s1 = iris.load_cube(str(DATA_DIR / "kgo_hi_res.nc"))
original_s1 = iris.load_cube(str(DATA_DIR / "original_algorithm_result.nc"))

print("温度:", temperature_meb.dims, temperature_meb.shape, "level=", list(temperature_meb.level.values))
print("地形:", topography_meb.dims, topography_meb.shape)
print("KGO:", kgo_s1.shape, kgo_s1.name())
````

### 3.2 输入数据绘图

气象场与 `MetaOrographicEnhancement` 默认 `boundary_height=1000` 对齐，展示 **1000 m** 层（样例里还有 100/200/500/1500 m，算法实际用的是最接近 1000 m 的那一层）。地形是二维场，没有高度维。

坐标是投影米制。官方地形文件在磁盘上 y 轴为北→南递减；`meb.read` 和算法内部会排成升序（南→北），与 KGO / 原方法结果一致。绘图里 `orient_for_plot` 若发现 y 递减会再翻一次，再配合 `ylim=(min, max)`，图面一律南在下、北在上，只改显示朝向，不改算法计算。


````python
# 与 MetaOrographicEnhancement(boundary_height=1000) 实际抽取的高度层一致
PLOT_LEVEL_M = 1000.0
plot_ref_met = to_plot_2d(select_level_for_plot(temperature_meb, PLOT_LEVEL_M))

scheme1_input_specs = [
    (temperature_meb, f"气温 ({PLOT_LEVEL_M:.0f} m)", "RdYlBu_r", lambda v: v - 273.15, "degC", True),
    (humidity_meb, f"相对湿度 ({PLOT_LEVEL_M:.0f} m)", "YlGnBu", lambda v: v * 100.0, "%", True),
    (pressure_meb, f"气压 ({PLOT_LEVEL_M:.0f} m)", "cividis", lambda v: v / 100.0, "hPa", True),
    (wind_speed_meb, f"风速 ({PLOT_LEVEL_M:.0f} m)", "Blues", None, "m s-1", True),
    (wind_direction_meb, f"风向 ({PLOT_LEVEL_M:.0f} m)", "twilight", None, "degrees", True),
    (topography_meb, "地形高度", "terrain", None, "m", False),
]

fig, axes = plt.subplots(2, 3, figsize=(16, 9), constrained_layout=True)
fig.suptitle("投影输入场", fontsize=15, y=1.02)
for ax, (field_meb, title, cmap, transform, unit, align_to_met) in zip(
    axes.ravel(), scheme1_input_specs, strict=True
):
    field_plot = to_plot_2d(select_level_for_plot(field_meb, PLOT_LEVEL_M)).copy()
    if transform is not None:
        field_plot.values = transform(np.asarray(field_plot.values))
    ref = plot_ref_met if align_to_met else None
    plot_field(ax, field_plot, title, reference=ref, cmap=cmap, norm=None, cbar_label=unit)
plt.show()
````

### 3.3 MetaOrographicEnhancement

先创建对象再传入 6 个场。输出在地形网格上。三个结果（迁移方法、原方法、KGO）应几乎一致。


````python
print("=== 投影：MetaOrographicEnhancement ===")
plugin_s1 = MetaOrographicEnhancement()
migrated_s1 = plugin_s1(
    temperature_meb,
    humidity_meb,
    pressure_meb,
    wind_speed_meb,
    wind_direction_meb,
    topography_meb,
)
migrated_s1.name = "orographic_enhancement"

compare_stats(migrated_s1, kgo_s1, "修改后方法 vs KGO")
compare_stats(migrated_s1, original_s1, "修改后方法 vs 原方法")
compare_stats(original_s1, kgo_s1, "原方法 vs KGO")
plot_precip_triplet(
    migrated_s1,
    kgo_s1,
    original_s1,
    title="投影 MetaOrographicEnhancement（修改后 | 原方法 | KGO）",
)
````

### 3.4 ApplyOrographicEnhancement

`precipitation.nc` 是已对齐到增强项网格的降水场（`lwe_precipitation_rate`，y 南→北）。
增强项用原方法结果 `original_algorithm_result.nc`（投影 meb 在 `cli_input/`）。

原方法读 Cube；迁移方法读 `cli_input/` 里的 meb。两者应几乎一致。


````python
require_files(
    [
        DATA_DIR / "precipitation.nc",
        DATA_DIR / "original_algorithm_result.nc",
        CLI_INPUT_DIR / "precipitation.nc",
        CLI_INPUT_DIR / "original_algorithm_result.nc",
    ],
    "ApplyOE 投影输入",
)

# 原方法：precipitation.nc 已与增强项同网格、同时刻、同 y 轴方向
precip_cube = iris.load_cube(str(DATA_DIR / "precipitation.nc"))
oe_cube = iris.load_cube(str(DATA_DIR / "original_algorithm_result.nc"))
original_applied = OriginalApplyOrographicEnhancement("add").process(
    iris.cube.CubeList([precip_cube]), oe_cube
)[0]

# 迁移方法：读预处理写出的 meb
precip_meb = meb.read_griddata_from_nc(str(CLI_INPUT_DIR / "precipitation.nc"))
oe_meb = meb.read_griddata_from_nc(str(CLI_INPUT_DIR / "original_algorithm_result.nc"))
migrated_applied = MigratedApplyOrographicEnhancement("add").process(precip_meb, oe_meb)

compare_stats(migrated_applied, original_applied, "ApplyOE 修改后方法 vs 原方法")

fig, axes = plt.subplots(1, 3, figsize=(22, 6), constrained_layout=True)
fig.suptitle("投影 ApplyOrographicEnhancement", fontsize=14, y=1.02)
plot_ref = to_plot_2d(precip_cube)
plot_field(axes[0], plot_ref, "参考降水场 (precipitation.nc)", reference=plot_ref)
mesh = plot_field(axes[1], to_plot_2d(migrated_applied), "迁移算法 ApplyOE", reference=plot_ref)
plot_field(axes[2], to_plot_2d(original_applied), "原算法 ApplyOE", reference=plot_ref)
add_precip_colorbar(fig, axes, mesh=mesh)
plt.show()
````

## 4. 经纬坐标输入

读取预处理写出的 `latlon/cli_input/` meb 与 `latlon/` Cube。
`latlon/` 下是经纬 Cube：ApplyOE 用 `precipitation.nc`、`original_algorithm_result.nc`，对照用重网格 `kgo_hi_res.nc`。气象/地形不写 Cube：原 `OrographicEnhancement` 不能在纯经纬上算梯度。
经纬网格和投影网格的格距不同，**不要拿投影 KGO 来比**。

- 计算增强项：只跑迁移方法，用重网格后的官方结果对照（KGO 读 Cube 即可，不必六维）。
- 叠加增强项：原 `ApplyOrographicEnhancement` **可以**在纯经纬上跑，下面会现场对照。


### 4.1 读取预处理结果


````python
latlon_meb_files = [
    LATLON_MEB_DIR / "temperature.nc",
    LATLON_MEB_DIR / "humidity.nc",
    LATLON_MEB_DIR / "pressure.nc",
    LATLON_MEB_DIR / "wind_speed.nc",
    LATLON_MEB_DIR / "wind_direction.nc",
    LATLON_MEB_DIR / "orography_uk-standard_1km.nc",
    LATLON_MEB_DIR / "original_algorithm_result.nc",
    LATLON_MEB_DIR / "precipitation.nc",
]
require_files(
    latlon_meb_files
    + [
        LATLON_DIR / "precipitation.nc",
        LATLON_DIR / "original_algorithm_result.nc",
        LATLON_DIR / "kgo_hi_res.nc",
    ],
    "经纬坐标输入",
)

temperature_ll = meb.read_griddata_from_nc(str(LATLON_MEB_DIR / "temperature.nc"))
humidity_ll = meb.read_griddata_from_nc(str(LATLON_MEB_DIR / "humidity.nc"))
pressure_ll = meb.read_griddata_from_nc(str(LATLON_MEB_DIR / "pressure.nc"))
wind_speed_ll = meb.read_griddata_from_nc(str(LATLON_MEB_DIR / "wind_speed.nc"))
wind_direction_ll = meb.read_griddata_from_nc(str(LATLON_MEB_DIR / "wind_direction.nc"))
topography_ll = meb.read_griddata_from_nc(str(LATLON_MEB_DIR / "orography_uk-standard_1km.nc"))
kgo_ll = iris.load_cube(str(LATLON_DIR / "kgo_hi_res.nc"))
original_ll = meb.read_griddata_from_nc(str(LATLON_MEB_DIR / "original_algorithm_result.nc"))

print("经纬温度:", temperature_ll.dims, temperature_ll.shape)
print("经纬地形:", topography_ll.dims, topography_ll.shape)
````

### 4.2 MetaOrographicEnhancement

主判据是同一经纬输入下「迁移方法 vs 重网格后的官方结果」。重网格 KGO 与原算法结果应几乎相同（它们来自同一套投影场）。


````python
print("=== 经纬：MetaOrographicEnhancement ===")
plugin_s2 = MetaOrographicEnhancement()
migrated_s2 = plugin_s2(
    temperature_ll,
    humidity_ll,
    pressure_ll,
    wind_speed_ll,
    wind_direction_ll,
    topography_ll,
)
migrated_s2.name = "orographic_enhancement"

compare_stats(migrated_s2, kgo_ll, "经纬 修改后方法 vs 重网格KGO")
compare_stats(migrated_s2, original_ll, "经纬 修改后方法 vs 重网格原方法")
compare_stats(original_ll, kgo_ll, "经纬 重网格原方法 vs 重网格KGO")
plot_precip_triplet(
    migrated_s2,
    kgo_ll,
    original_ll,
    title="经纬 MetaOrographicEnhancement（修改后 | 重网格原方法 | 重网格KGO）",
)
````

### 4.3 ApplyOrographicEnhancement

降水与增强项已在同一张经纬网格上（预处理时 lat 南→北、lon 西→东）。
增强项用原方法结果 `original_algorithm_result.nc`。
原方法只读 `latlon/` 下这两份 Cube；迁移方法读 `latlon/cli_input/` meb。两者应几乎一致。


````python
precip_ll_cube = iris.load_cube(str(LATLON_DIR / "precipitation.nc"))
oe_ll_cube = iris.load_cube(str(LATLON_DIR / "original_algorithm_result.nc"))
original_applied_ll = OriginalApplyOrographicEnhancement("add").process(
    iris.cube.CubeList([precip_ll_cube]), oe_ll_cube
)[0]

precip_ll_meb = meb.read_griddata_from_nc(str(LATLON_MEB_DIR / "precipitation.nc"))
oe_ll_meb = meb.read_griddata_from_nc(str(LATLON_MEB_DIR / "original_algorithm_result.nc"))
migrated_applied_ll = MigratedApplyOrographicEnhancement("add").process(
    precip_ll_meb, oe_ll_meb
)

compare_stats(migrated_applied_ll, original_applied_ll, "经纬 ApplyOE 修改后方法 vs 原方法")

fig, axes = plt.subplots(1, 3, figsize=(22, 6), constrained_layout=True)
fig.suptitle("经纬 ApplyOrographicEnhancement", fontsize=14, y=1.02)
plot_ref = to_plot_2d(precip_ll_cube)
plot_field(axes[0], plot_ref, "参考降水场", reference=plot_ref)
mesh = plot_field(axes[1], to_plot_2d(migrated_applied_ll), "迁移算法 ApplyOE", reference=plot_ref)
plot_field(axes[2], to_plot_2d(original_applied_ll), "原算法 ApplyOE", reference=plot_ref)
add_precip_colorbar(fig, axes, mesh=mesh)
plt.show()
````
