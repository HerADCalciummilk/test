# Precipitation transform

Run from this package directory. Omitted arguments are read from `resource/db_trans.ini`, then code presets.

```bash
python -m cli.dB_trans_cli
python -m cli.dB_trans_cli --method db --values 0,0.05,0.1,1,10,100
python -m cli.dB_trans_cli --method box --values 0.1,1,10 --lambda 0
```

With method `db` and threshold `0.1`, `0,0.05,0.1,1,10,100` becomes `-15,-15,-10,0,10,20`.


````python
import numpy as np
from src.dB_trans import dB_transform

values = np.array([0.0, 0.05, 0.1, 1.0, 10.0, 100.0])
result, metadata = dB_transform(values)
print(result)
print(metadata["transform"], metadata["threshold"], metadata["zerovalue"])
````
