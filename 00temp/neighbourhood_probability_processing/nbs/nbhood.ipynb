**预处理不在本 Notebook 中执行。** 请先在仓库根目录运行：

```bash

```

预处理约定：官方投影维仅重命名为 `lat`/`lon`（数值仍为投影米制）；**阈值维 `threshold` 映射到六维 `level`**；`realization` 映射到 `member`；真经纬输入写出到各场景 `cli_input/.../latlon/`。

对比图为：预处理输入、修改后算法、KGO、原算法结果。三者两两做差，打印平均绝对误差、最大绝对误差和均方根误差，不再画差值图。

插件场景另附**经纬输入验证**（读 `latlon/` 真重网格 meb）：跑迁移版经纬→LAEA 路径，确认坐标恢复；并与预处理写出的「投影路径结果再重网格到同一经纬」对照场（`ref_projected_*.nc`）做**有界对照**（Notebook 只读盘；不要求差为 0，不与官方投影 KGO 硬比）。经纬验证只放在插件场景，不在 CLI 段重复。
LAEA 接线自洽（经纬完整路径 ≡ 适配米轴投影路径）由 `pytest` 覆盖。

## 对比工具

绘图、member 选择，以及修改后 / 原方法 / KGO 三者数值差异打印。

- Cube：`to_plot_2d` / `field_2d_for_compare` 保留 Iris MaskedArray
- meb / CLI 返回值：只对真正的 `NaN` 做 `masked_invalid`
- `mask_fill_values_array` + `FILL_THRESHOLD` 仅用于内部掩码输入（`input_masked.nc`）


````python
def _detect_spatial_dims(da):
    for y_name, x_name in (
        ("lat", "lon"),
        ("latitude", "longitude"),
        ("projection_y_coordinate", "projection_x_coordinate"),
    ):
        if y_name in da.dims and x_name in da.dims:
            return y_name, x_name
    return da.dims[-2], da.dims[-1]


def _is_iris_cube(obj) -> bool:
    return (
        not isinstance(obj, (xr.DataArray, np.ndarray, np.ma.MaskedArray))
        and hasattr(obj, "coord")
        and hasattr(obj, "data")
    )


def _iris_cube_to_2d(cube, member_index=0, level_index=0):
    """把官方 Iris Cube 压到二维（最后两维为空间）。"""
    spatial_names = {
        "projection_y_coordinate",
        "projection_x_coordinate",
        "latitude",
        "longitude",
        "grid_latitude",
        "grid_longitude",
    }
    work = cube
    while work.ndim > 2:
        dim_name = work.dim_coords[0].name() if work.dim_coords else None
        if dim_name in spatial_names:
            break
        if dim_name in ("realization", "member"):
            idx = min(member_index, work.shape[0] - 1)
        elif dim_name in ("threshold", "level", "percentile"):
            idx = min(level_index, work.shape[0] - 1)
        else:
            idx = 0
        work = work[idx]
    raw = work.data
    if np.ma.isMaskedArray(raw):
        return np.squeeze(np.ma.asarray(raw))
    return np.squeeze(np.asarray(raw, dtype=np.float64))


def to_plot_2d(da, member_index=0, level_index=0):
    if _is_iris_cube(da):
        return _iris_cube_to_2d(da, member_index=member_index, level_index=level_index)
    if not isinstance(da, xr.DataArray):
        # MaskedArray 需保留掩码位（内部掩码 numpy 路径的有效计算靠 mask 表达）。
        arr = np.ma.asarray(da) if np.ma.isMaskedArray(da) else np.asarray(da)
        return arr.squeeze() if arr.ndim > 2 else arr
    work = da
    y_name, x_name = _detect_spatial_dims(work)
    if "member" in work.dims:
        work = work.isel(member=min(member_index, work.sizes["member"] - 1))
    if "level" in work.dims:
        work = work.isel(level=min(level_index, work.sizes["level"] - 1))
    indexers = {d: 0 for d in work.dims if d not in (y_name, x_name)}
    if indexers:
        work = work.isel(indexers)
    return np.asarray(work.values)


def member_groups_by_percentile(da):
    if not isinstance(da, xr.DataArray):
        return {}
    if da.attrs.get("level_type") == "neighbourhood_percentile" and "level" in da.dims:
        return {float(v): list(range(da.sizes.get("member", 1))) for v in da.coords["level"].values}
    if "member" not in da.dims:
        return {}
    key = next(
        (
            cand
            for cand in ("member_percentile", "percentile")
            if cand in da.coords and da.coords[cand].dims == ("member",)
        ),
        None,
    )
    if key is None:
        return {}
    groups = {}
    for i, v in enumerate(np.asarray(da.coords[key].values, dtype=np.float64)):
        groups.setdefault(float(v), []).append(i)
    return groups


def pick_member_by_percentile(da, percentile_value, prefer_realization_index=0):
    if isinstance(da, xr.DataArray) and da.attrs.get("level_type") == "neighbourhood_percentile":
        return min(prefer_realization_index, da.sizes.get("member", 1) - 1)
    groups = member_groups_by_percentile(da)
    if not groups:
        return 0
    p = float(percentile_value)
    if p not in groups:
        return groups[sorted(groups.keys())[0]][0]
    candidates = groups[p]
    return int(candidates[min(prefer_realization_index, len(candidates) - 1)])


def to_percentile_2d(da, percentile_value, prefer_realization_index=0):
    """提取指定百分位的二维场（兼容 Iris Cube、level 上的邻域百分位或 realization/percentile 维）。"""
    if _is_iris_cube(da):
        cube = da
        if cube.coords("percentile"):
            points = np.asarray(cube.coord("percentile").points, dtype=np.float64)
            pdim = cube.coord_dims("percentile")
            if pdim:
                sl = [slice(None)] * cube.ndim
                sl[pdim[0]] = int(np.argmin(np.abs(points - float(percentile_value))))
                cube = cube[tuple(sl)]
        return to_plot_2d(cube, member_index=prefer_realization_index)
    if isinstance(da, xr.DataArray) and da.attrs.get("level_type") == "neighbourhood_percentile":
        work = da
        if "member" in work.dims:
            work = work.isel(member=min(prefer_realization_index, work.sizes["member"] - 1))
        work = work.sel(level=float(percentile_value), method="nearest")
        y_name, x_name = _detect_spatial_dims(work)
        indexers = {d: 0 for d in work.dims if d not in (y_name, x_name)}
        if indexers:
            work = work.isel(indexers)
        return np.asarray(work.transpose(y_name, x_name).values)
    if isinstance(da, xr.DataArray) and "percentile" in da.dims and "member" not in da.dims:
        work = da
        if "realization" in work.dims:
            rvals = np.asarray(work.coords["realization"].values)
            ridx = min(prefer_realization_index, len(rvals) - 1)
            work = work.sel(realization=rvals[ridx])
        work = work.sel(percentile=float(percentile_value), method="nearest")
        y_name, x_name = _detect_spatial_dims(work)
        indexers = {d: 0 for d in work.dims if d not in (y_name, x_name)}
        if indexers:
            work = work.isel(indexers)
        return np.asarray(work.transpose(y_name, x_name).values)
    idx = pick_member_by_percentile(da, percentile_value, prefer_realization_index)
    return to_plot_2d(da, member_index=idx)


def field_2d_for_compare(da, *, member_index=0, level_index=0):
    """将 CLI/插件结果压成可对比的二维场（阈值在 level；兼容旧 threshold 维）。

    Cube 走 ``to_plot_2d`` 已保留 Iris MaskedArray；meb 结果只把 ``NaN`` 标成缺测。
    """
    work = da
    if isinstance(work, xr.DataArray) and "threshold" in work.dims:
        work = work.isel(threshold=min(level_index, work.sizes["threshold"] - 1))
    field = to_plot_2d(work, member_index=member_index, level_index=level_index)
    if np.ma.isMaskedArray(field):
        return field
    return np.ma.masked_invalid(np.asarray(field, dtype=np.float64))


def compare_stats(current, reference, tag: str) -> None:
    """打印两个场的差异。三个数都接近 0，说明两边几乎一样。"""
    left = np.ma.filled(np.ma.asarray(current, dtype=np.float64), np.nan)
    right = np.ma.filled(np.ma.asarray(reference, dtype=np.float64), np.nan)
    left = np.squeeze(left)
    right = np.squeeze(right)
    if left.shape != right.shape:
        raise ValueError(f"形状不一致: {left.shape} vs {right.shape}（{tag}）")
    valid = np.isfinite(left) & np.isfinite(right)
    if not np.any(valid):
        print(f"[{tag}] 无有效格点可比")
        return
    diff = left[valid] - right[valid]
    print(
        f"[{tag}] "
        f"平均绝对误差={np.mean(np.abs(diff)):.8f}, "
        f"最大绝对误差={np.max(np.abs(diff)):.8f}, "
        f"均方根误差={np.sqrt(np.mean(diff ** 2)):.8f}"
    )


def print_triplet_diffs(modified, kgo, original, prefix: str) -> None:
    """修改后方法、原方法、KGO 三者两两做差并打印。"""
    compare_stats(modified, kgo, f"{prefix} 修改后方法 vs KGO")
    compare_stats(modified, original, f"{prefix} 修改后方法 vs 原方法")
    compare_stats(original, kgo, f"{prefix} 原方法 vs KGO")


def mask_fill_values_array(data, fill_threshold=FILL_THRESHOLD):
    """大填充值转为 MaskedArray（算法内部掩码路径用）。"""
    values = np.asarray(
        data.values if isinstance(data, xr.DataArray) else data,
        dtype=np.float32,
    )
    masked = np.ma.masked_invalid(values)
    return np.ma.masked_where(
        np.abs(np.ma.filled(masked, 0.0)) >= fill_threshold,
        masked,
    )


def common_limits(*arrays):
    mins, maxs = [], []
    for array in arrays:
        data = np.ma.asarray(array).filled(np.nan)
        mins.append(float(np.nanmin(data)))
        maxs.append(float(np.nanmax(data)))
    return min(mins), max(maxs)


def plot_basic(title, input_data, modified, kgo, original):
    print_triplet_diffs(modified, kgo, original, title)
    vmin, vmax = common_limits(modified, kgo, original)
    fig, axes = plt.subplots(1, 4, figsize=(22, 5))
    panels = [
        (input_data, "原输入", None),
        (modified, "修改后算法", (vmin, vmax)),
        (kgo, "KGO", (vmin, vmax)),
        (original, "原算法", (vmin, vmax)),
    ]
    for ax, (data, name, limits) in zip(axes, panels):
        kwargs = {} if limits is None else {"vmin": limits[0], "vmax": limits[1]}
        im = ax.imshow(data, cmap="RdYlBu_r", interpolation="nearest", **kwargs)
        ax.set_title(name)
        plt.colorbar(im, ax=ax, fraction=0.046)
    fig.suptitle(title, fontsize=16)
    fig.tight_layout()
    plt.show()


def plot_cli_triplet(title, cli_modified, kgo, cli_original, *, member_index=0, level_index=0):
    """修改后 CLI、KGO、原 CLI 并图对比；三者两两做差打印数值差异。"""
    panels = [
        (
            field_2d_for_compare(
                cli_modified, member_index=member_index, level_index=level_index
            ),
            "修改后 CLI",
        ),
        (
            field_2d_for_compare(kgo, member_index=member_index, level_index=level_index),
            "KGO",
        ),
        (
            field_2d_for_compare(
                cli_original, member_index=member_index, level_index=level_index
            ),
            "原 CLI",
        ),
    ]
    print_triplet_diffs(panels[0][0], panels[1][0], panels[2][0], title)
    vmin, vmax = common_limits(*[data for data, _ in panels])
    fig, axes = plt.subplots(1, 3, figsize=(18, 5))
    fig.suptitle(title, fontsize=16)
    for ax, (data, label) in zip(axes, panels):
        im = ax.imshow(data, cmap="RdYlBu_r", vmin=vmin, vmax=vmax, interpolation="nearest")
        ax.set_title(label)
        plt.colorbar(im, ax=ax, fraction=0.046)
    fig.tight_layout()
    plt.show()


def plot_cli_percentile_triplet(title, cli_modified, kgo, cli_original):
    """百分位结果：每一百分位一行（修改后 CLI / KGO / 原 CLI）。"""
    fig, axes = plt.subplots(
        len(PERCENTILES), 3, figsize=(18, 5 * len(PERCENTILES)), squeeze=False
    )
    col_labels = ("修改后 CLI", "KGO", "原 CLI")
    for row, pct in enumerate(PERCENTILES):
        slices = [
            to_percentile_2d(cli_modified, pct),
            to_percentile_2d(kgo, pct),
            to_percentile_2d(cli_original, pct),
        ]
        vmin, vmax = common_limits(*slices)
        print_triplet_diffs(slices[0], slices[1], slices[2], f"{title} {pct:.0f}%")
        for col, (data, label) in enumerate(zip(slices, col_labels)):
            ax = axes[row, col]
            im = ax.imshow(
                data, cmap="RdYlBu_r", vmin=vmin, vmax=vmax, interpolation="nearest"
            )
            ax.set_title(f"{label}（{pct:.0f}%）" if col == 0 else f"{pct:.0f}%")
            plt.colorbar(im, ax=ax, fraction=0.046)
    fig.suptitle(title, fontsize=16)
    fig.tight_layout()
    plt.show()


MASK_COVER_CMAP = ListedColormap(["#ffffff", "#404040"])
MASK_COVER_NORM = BoundaryNorm([-0.5, 0.5, 1.5], MASK_COVER_CMAP.N)


def plot_mask(title, cover, modified, kgo, original, *, cover_label="有效区域 (1=参与)"):
    """cover：0/1 二值覆盖图；1 的含义由 cover_label 说明。"""
    print_triplet_diffs(modified, kgo, original, title)
    vmin, vmax = common_limits(modified, kgo, original)
    fig, axes = plt.subplots(1, 4, figsize=(22, 5))
    cover_01 = np.asarray(cover, dtype=np.uint8)
    im = axes[0].imshow(
        cover_01,
        cmap=MASK_COVER_CMAP,
        norm=MASK_COVER_NORM,
        interpolation="nearest",
    )
    axes[0].set_title(cover_label)
    cbar = plt.colorbar(im, ax=axes[0], fraction=0.046, ticks=[0, 1])
    cbar.ax.set_yticklabels(["0", "1"])
    for ax, data, name in [
        (axes[1], modified, "修改后"),
        (axes[2], kgo, "KGO"),
        (axes[3], original, "原算法"),
    ]:
        im = ax.imshow(
            data, cmap="RdYlBu_r", vmin=vmin, vmax=vmax, interpolation="nearest"
        )
        ax.set_title(name)
        plt.colorbar(im, ax=ax, fraction=0.046)
    fig.suptitle(title, fontsize=16)
    fig.tight_layout()
    plt.show()

import json
from pyproj import CRS, Transformer

from neighbourhood_probability_processing.src.utils._regrid import is_geographic_spatial_dataarray


def assert_geographic_coords(result: xr.DataArray, template: xr.DataArray, tag: str) -> None:
    """确认经纬路径输出坐标恢复为输入经纬模板。"""
    if not isinstance(result, xr.DataArray):
        raise TypeError(f"{tag}: 期望 DataArray，得到 {type(result)}")
    if not is_geographic_spatial_dataarray(result):
        raise AssertionError(f"{tag}: 输出仍非经纬坐标")
    if not np.allclose(result.lat.values, template.lat.values):
        raise AssertionError(f"{tag}: lat 未恢复为输入模板")
    if not np.allclose(result.lon.values, template.lon.values):
        raise AssertionError(f"{tag}: lon 未恢复为输入模板")
    print(
        f"[{tag}] 经纬路径坐标恢复通过：shape={tuple(result.shape)}, "
        f"lat units={result.lat.attrs.get('units')}, "
        f"lon units={result.lon.attrs.get('units')}"
    )


def plot_latlon_bounded(
    title: str,
    geo_input,
    geo_result,
    proj_on_latlon,
    *,
    level_index: int = 0,
    mae_warn: float = 0.25,
) -> None:
    """经纬输入 / 经纬结果 / 预处理写出的投影结果重网格场；有界对照。"""
    left = field_2d_for_compare(geo_input, level_index=level_index)
    mid = field_2d_for_compare(geo_result, level_index=level_index)
    right = field_2d_for_compare(proj_on_latlon, level_index=level_index)
    compare_stats(mid, right, f"{title} 经纬结果 vs 投影结果重网格")
    a = np.ma.filled(np.ma.asarray(mid, dtype=np.float64), np.nan)
    b = np.ma.filled(np.ma.asarray(right, dtype=np.float64), np.nan)
    valid = np.isfinite(a) & np.isfinite(b)
    if np.any(valid):
        mae = float(np.mean(np.abs(a[valid] - b[valid])))
        print(
            f"[{title}] 有界对照 MAE={mae:.6f}"
            + (f"（提示：超过参考阈值 {mae_warn}）" if mae > mae_warn else "（量级可接受）")
        )
    vmin, vmax = common_limits(left, mid, right)
    fig, axes = plt.subplots(1, 3, figsize=(18, 5))
    for ax, data, name in (
        (axes[0], left, "经纬输入"),
        (axes[1], mid, "经纬路径结果"),
        (axes[2], right, "投影结果→经纬（预处理）"),
    ):
        im = ax.imshow(
            data, cmap="RdYlBu_r", vmin=vmin, vmax=vmax, interpolation="nearest"
        )
        ax.set_title(name)
        plt.colorbar(im, ax=ax, fraction=0.046)
    fig.suptitle(title, fontsize=16)
    fig.tight_layout()
    plt.show()
````

## 掩码邻域

掩码处理分两种情形验证：

- **外部掩码**：预处理已写出 `cli_input/mask/input.nc` 与 `cli_input/mask/mask.nc`；`mask=1` 为参与邻域的有效格点。
- **内部掩码**：预处理已写出 `cli_input/mask/input_masked.nc`（保留大填充值）；调用前转为 `MaskedArray`，以 `grid_spacing=2000 m` 走 numpy 路径。

因 xr.DataArray 输入无法携带内部掩码，故内部掩码验证使用 MaskedArray 作为输入。

输出语义（`re_mask=True`，对输出重掩码）：

- **外部掩码 + `DataArray`**：无效格点写为 `NaN`，有效格点为邻域统计结果。CLI 以 float32 `to_netcdf`（`_FillValue=NaN`）写出；CLI 验证单元直接用 `process()` 返回值对比，避免再读盘占用输出文件。
- **内部掩码 + `MaskedArray`**：返回 `MaskedArray`，无效位以 `mask` 标记（底层 `.data` 仍是邻域统计值），与原方法一致。
- 官方 KGO / 原结果用 `iris.load_cube` 读，对比时保留 Iris MaskedArray。

## 百分位验证

圆形邻域百分位输出，对比 KGO 与原算法。每一百分位打印三者两两数值差异，不再画差值图。


````python
pct_dir = DATA_DIR / "percentile"

pct_modified = GeneratePercentilesFromANeighbourhood(
    RADIUS_METRES,
    percentiles=PERCENTILES,
).process(pct_input)
pct_kgo = iris.load_cube(str(pct_dir / "kgo_circular_percentile.nc"))
pct_original = iris.load_cube(str(pct_dir / "original_circular_percentile.nc"))

fig, axes = plt.subplots(len(PERCENTILES), 3, figsize=(18, 5 * len(PERCENTILES)), squeeze=False)
for row, pct in enumerate(PERCENTILES):
    m = to_percentile_2d(pct_modified, pct, prefer_realization_index=0)
    k = to_percentile_2d(pct_kgo, pct, prefer_realization_index=0)
    o = to_percentile_2d(pct_original, pct, prefer_realization_index=0)
    print_triplet_diffs(m, k, o, f"百分位数 {pct:.0f}%")
    vmin, vmax = common_limits(m, k, o)
    for ax, data, name in [
        (axes[row, 0], m, f"修改后 {pct:.0f}th"),
        (axes[row, 1], k, f"KGO {pct:.0f}th"),
        (axes[row, 2], o, f"原算法 {pct:.0f}th"),
    ]:
        im = ax.imshow(data, cmap="RdYlBu_r", vmin=vmin, vmax=vmax, interpolation="nearest")
        ax.set_title(name)
        plt.colorbar(im, ax=ax, fraction=0.046)
fig.suptitle("百分位数对比", fontsize=16)
fig.tight_layout()
plt.show()
````
