## 2. 数据读取与绘图工具

- 预处理 `cli_input/`：单变量 nc，用 `xr.open_dataarray`（掩码/权重为六维 meb（带维 `level`），可用 `meb.read_griddata_from_nc`）。
- 修改后 CLI 结果：直接用 `process()` 返回值（写出 float32、`_FillValue=NaN`），避免再读盘占用输出文件。
- 官方 KGO / 原算法 / 原 CLI：`iris.load_cube`，对比时保留 Iris MaskedArray，不再用 `1e20` 拆开再打 mask。

结果图中的修改后、KGO、原结果三幅图统一颜色刻度。

````python
def _detect_spatial_dims(da):
    for y_name, x_name in (
        ("lat", "lon"),
        ("latitude", "longitude"),
        ("projection_y_coordinate", "projection_x_coordinate"),
    ):
        if y_name in da.dims and x_name in da.dims:
            return y_name, x_name
    return da.dims[-2], da.dims[-1]


def _is_iris_cube(obj) -> bool:
    return (
        not isinstance(obj, (xr.DataArray, np.ndarray, np.ma.MaskedArray))
        and hasattr(obj, "coord")
        and hasattr(obj, "data")
    )


def _iris_cube_to_2d(cube, member_index=0, level_index=0, zone_index=0):
    """把官方 Iris Cube 压到二维（最后两维为空间）。"""
    spatial_names = {
        "projection_y_coordinate",
        "projection_x_coordinate",
        "latitude",
        "longitude",
        "grid_latitude",
        "grid_longitude",
    }
    work = cube
    while work.ndim > 2:
        dim_name = work.dim_coords[0].name() if work.dim_coords else None
        if dim_name in spatial_names:
            break
        if dim_name in ("realization", "member"):
            idx = min(member_index, work.shape[0] - 1)
        elif dim_name in ("threshold", "level", "percentile"):
            idx = min(level_index, work.shape[0] - 1)
        elif dim_name in ("level", "topographic_zone"):
            idx = min(zone_index, work.shape[0] - 1)
        else:
            idx = 0
        work = work[idx]
    raw = work.data
    if np.ma.isMaskedArray(raw):
        return np.squeeze(np.ma.asarray(raw))
    return np.squeeze(np.asarray(raw, dtype=np.float64))


def to_plot_2d(da, member_index=0, level_index=0, zone_index=0):
    if _is_iris_cube(da):
        return _iris_cube_to_2d(
            da, member_index=member_index, level_index=level_index, zone_index=zone_index
        )
    if not isinstance(da, xr.DataArray):
        arr = np.ma.asarray(da) if np.ma.isMaskedArray(da) else np.asarray(da)
        return arr.squeeze() if arr.ndim > 2 else arr
    work = da
    y_name, x_name = _detect_spatial_dims(work)
    if "member" in work.dims:
        work = work.isel(member=min(member_index, work.sizes["member"] - 1))
    # 带维亦为 level 时留给下方 COORD_FOR_MASKING（zone_index）选取
    if "level" in work.dims and COORD_FOR_MASKING != "level":
        work = work.isel(level=min(level_index, work.sizes["level"] - 1))
    if "threshold" in work.dims:
        work = work.isel(threshold=min(level_index, work.sizes["threshold"] - 1))
    if COORD_FOR_MASKING in work.dims:
        work = work.isel(
            **{COORD_FOR_MASKING: min(zone_index, work.sizes[COORD_FOR_MASKING] - 1)}
        )
    elif work.attrs.get("is_threshold_zone_stacked") == "True" and "level" in work.dims:
        # 未折叠：阈值×地形带已并入 level；优先按 topographic_zone 辅助坐标选带层
        if "topographic_zone" in work.coords:
            zones = np.unique(np.asarray(work.coords["topographic_zone"].values))
            work = work.sel(topographic_zone=zones[min(zone_index, zones.size - 1)])
        else:
            work = work.isel(level=min(zone_index, work.sizes["level"] - 1))
    indexers = {d: 0 for d in work.dims if d not in (y_name, x_name)}
    if indexers:
        work = work.isel(indexers)
    return np.asarray(work.values)


def field_2d_for_compare(da, *, member_index=0, level_index=0, zone_index=0):
    """压成可对比二维场。Cube 保留 Iris MaskedArray；meb 结果只把 NaN 标成缺测。"""
    field = to_plot_2d(
        da, member_index=member_index, level_index=level_index, zone_index=zone_index
    )
    if np.ma.isMaskedArray(field):
        return field
    return np.ma.masked_invalid(np.asarray(field, dtype=np.float64))


def compare_stats(current, reference, tag: str) -> None:
    left = np.ma.filled(np.ma.asarray(current, dtype=np.float64), np.nan)
    right = np.ma.filled(np.ma.asarray(reference, dtype=np.float64), np.nan)
    left = np.squeeze(left)
    right = np.squeeze(right)
    if left.shape != right.shape:
        raise ValueError(f"形状不一致: {left.shape} vs {right.shape}（{tag}）")
    valid = np.isfinite(left) & np.isfinite(right)
    if not np.any(valid):
        print(f"[{tag}] 无有效格点可比")
        return
    diff = left[valid] - right[valid]
    print(
        f"[{tag}] "
        f"平均绝对误差={np.mean(np.abs(diff)):.8f}, "
        f"最大绝对误差={np.max(np.abs(diff)):.8f}, "
        f"均方根误差={np.sqrt(np.mean(diff ** 2)):.8f}"
    )


def print_triplet_diffs(modified, kgo, original, prefix: str) -> None:
    compare_stats(modified, kgo, f"{prefix} 修改后方法 vs KGO")
    compare_stats(modified, original, f"{prefix} 修改后方法 vs 原方法")
    compare_stats(original, kgo, f"{prefix} 原方法 vs KGO")


def common_limits(*arrays):
    """计算多个数组共同使用的颜色刻度范围。"""
    mins, maxs = [], []
    for array in arrays:
        data = np.ma.asarray(array).filled(np.nan)
        mins.append(float(np.nanmin(data)))
        maxs.append(float(np.nanmax(data)))
    return min(mins), max(maxs)


def plot_auxiliary_data(input_data, mask_data, weights_data, threshold_index=0):
    """绘制输入场、地形带掩码和地形带权重（阈值优先取 level）。"""
    fig, axes = plt.subplots(2, 4, figsize=(20, 10))
    if "level" in input_data.dims and input_data.sizes.get("level", 1) > 1:
        input_slice = input_data.isel(level=threshold_index)
    elif "member" in input_data.dims:
        input_slice = input_data.isel(member=threshold_index)
    else:
        input_slice = input_data
    if isinstance(input_slice, xr.DataArray):
        for dim in ("level", "time", "dtime"):
            if dim in input_slice.dims:
                input_slice = input_slice.isel({dim: 0})
        input_slice = input_slice.squeeze(drop=True)
    input_values = np.squeeze(np.asarray(input_slice))
    if input_values.ndim != 2:
        raise ValueError(f"输入场无法绘制为二维图像: shape={input_values.shape}")
    im = axes[0, 0].imshow(input_values, origin="lower", cmap="RdYlBu_r", interpolation="nearest")
    axes[0, 0].set_title(f"输入概率场 level/threshold index={threshold_index}")
    plt.colorbar(im, ax=axes[0, 0], fraction=0.046)

    for index in range(mask_data.sizes[COORD_FOR_MASKING]):
        ax = axes[0, index + 1]
        mask_slice = mask_data.isel({COORD_FOR_MASKING: index})
        if isinstance(mask_slice, xr.DataArray):
            for dim in ("member", "level", "time", "dtime"):
                if dim in mask_slice.dims:
                    mask_slice = mask_slice.isel({dim: 0})
            mask_slice = mask_slice.squeeze(drop=True)
        mask_values = np.squeeze(np.asarray(mask_slice))
        if mask_values.ndim != 2:
            raise ValueError(f"掩码层无法绘制为二维图像: shape={mask_values.shape}")
        im = ax.imshow(mask_values, origin="lower", cmap="gray", vmin=0, vmax=1, interpolation="nearest")
        ax.set_title(f"掩码层 {index}")
        plt.colorbar(im, ax=ax, fraction=0.046)

    axes[1, 0].axis("off")
    for index in range(weights_data.sizes[COORD_FOR_MASKING]):
        ax = axes[1, index + 1]
        weight_slice = weights_data.isel({COORD_FOR_MASKING: index})
        if isinstance(weight_slice, xr.DataArray):
            for dim in ("member", "level", "time", "dtime"):
                if dim in weight_slice.dims:
                    weight_slice = weight_slice.isel({dim: 0})
            weight_slice = weight_slice.squeeze(drop=True)
        weight_values = np.squeeze(np.asarray(weight_slice))
        if weight_values.ndim != 2:
            raise ValueError(f"权重层无法绘制为二维图像: shape={weight_values.shape}")
        im = ax.imshow(weight_values, origin="lower", cmap="viridis", interpolation="nearest")
        ax.set_title(f"权重层 {index}")
        plt.colorbar(im, ax=ax, fraction=0.046)

    fig.suptitle("辅助数据图", fontsize=16)
    fig.tight_layout()
    plt.show()


def plot_result_comparison(
    title, modified, kgo, original, *, member_index=0, level_index=None, zone_index=0
):
    """绘制修改后算法、KGO 和原算法结果对比图。"""
    if level_index is None:
        level_index = THRESHOLD_INDEX
    slices = [
        field_2d_for_compare(
            modified, member_index=member_index, level_index=level_index, zone_index=zone_index
        ),
        field_2d_for_compare(
            kgo, member_index=member_index, level_index=level_index, zone_index=zone_index
        ),
        field_2d_for_compare(
            original, member_index=member_index, level_index=level_index, zone_index=zone_index
        ),
    ]
    print_triplet_diffs(slices[0], slices[1], slices[2], title)
    vmin, vmax = common_limits(*slices)
    fig, axes = plt.subplots(1, 3, figsize=(18, 5))
    panels = [
        (slices[0], "修改后算法结果"),
        (slices[1], "官方 KGO 结果"),
        (slices[2], "原算法结果"),
    ]
    for ax, (data, name) in zip(axes, panels):
        im = ax.imshow(data, origin="lower", cmap="RdYlBu_r", vmin=vmin, vmax=vmax, interpolation="nearest")
        ax.set_title(name)
        plt.colorbar(im, ax=ax, fraction=0.046)
    fig.suptitle(title, fontsize=16)
    fig.tight_layout()
    plt.show()


def plot_cli_triplet(
    title, cli_modified, kgo, cli_original, *, member_index=0, level_index=None, zone_index=0
):
    """修改后 CLI、KGO、原 CLI 并图对比。"""
    if level_index is None:
        level_index = THRESHOLD_INDEX
    slices = [
        field_2d_for_compare(
            cli_modified, member_index=member_index, level_index=level_index, zone_index=zone_index
        ),
        field_2d_for_compare(
            kgo, member_index=member_index, level_index=level_index, zone_index=zone_index
        ),
        field_2d_for_compare(
            cli_original, member_index=member_index, level_index=level_index, zone_index=zone_index
        ),
    ]
    print_triplet_diffs(slices[0], slices[1], slices[2], title)
    vmin, vmax = common_limits(*slices)
    fig, axes = plt.subplots(1, 3, figsize=(18, 5))
    labels = ("修改后 CLI", "KGO", "原 CLI")
    for ax, (data, label) in zip(axes, zip(slices, labels)):
        im = ax.imshow(data, origin="lower", cmap="RdYlBu_r", vmin=vmin, vmax=vmax, interpolation="nearest")
        ax.set_title(label)
        plt.colorbar(im, ax=ax, fraction=0.046)
    fig.suptitle(title, fontsize=16)
    fig.tight_layout()
    plt.show()

import json
from pyproj import CRS, Transformer

from neighbourhood_probability_processing.src.utils._regrid import is_geographic_spatial_dataarray


def assert_geographic_coords(result: xr.DataArray, template: xr.DataArray, tag: str) -> None:
    """确认经纬路径输出坐标恢复为输入经纬模板。"""
    if not isinstance(result, xr.DataArray):
        raise TypeError(f"{tag}: 期望 DataArray，得到 {type(result)}")
    if not is_geographic_spatial_dataarray(result):
        raise AssertionError(f"{tag}: 输出仍非经纬坐标")
    if not np.allclose(result.lat.values, template.lat.values):
        raise AssertionError(f"{tag}: lat 未恢复为输入模板")
    if not np.allclose(result.lon.values, template.lon.values):
        raise AssertionError(f"{tag}: lon 未恢复为输入模板")
    print(
        f"[{tag}] 经纬路径坐标恢复通过：shape={tuple(result.shape)}, "
        f"lat units={result.lat.attrs.get('units')}, "
        f"lon units={result.lon.attrs.get('units')}"
    )


def plot_latlon_bounded(
    title: str,
    geo_input,
    geo_result,
    proj_on_latlon,
    *,
    level_index: int = 0,
    zone_index: int = 0,
    mae_warn: float = 0.25,
) -> None:
    """经纬输入 / 经纬结果 / 预处理写出的投影结果重网格场；有界对照。"""
    left = field_2d_for_compare(
        geo_input, level_index=level_index, zone_index=zone_index
    )
    mid = field_2d_for_compare(
        geo_result, level_index=level_index, zone_index=zone_index
    )
    right = field_2d_for_compare(
        proj_on_latlon, level_index=level_index, zone_index=zone_index
    )
    compare_stats(mid, right, f"{title} 经纬结果 vs 投影结果重网格")
    a = np.ma.filled(np.ma.asarray(mid, dtype=np.float64), np.nan)
    b = np.ma.filled(np.ma.asarray(right, dtype=np.float64), np.nan)
    valid = np.isfinite(a) & np.isfinite(b)
    if np.any(valid):
        mae = float(np.mean(np.abs(a[valid] - b[valid])))
        print(
            f"[{title}] 有界对照 MAE={mae:.6f}"
            + (f"（提示：超过参考阈值 {mae_warn}）" if mae > mae_warn else "（量级可接受）")
        )
    vmin, vmax = common_limits(left, mid, right)
    fig, axes = plt.subplots(1, 3, figsize=(18, 5))
    for ax, data, name in (
        (axes[0], left, "经纬输入"),
        (axes[1], mid, "经纬路径结果"),
        (axes[2], right, "投影结果→经纬（预处理）"),
    ):
        im = ax.imshow(
            data,
            origin="lower",
            cmap="RdYlBu_r",
            vmin=vmin,
            vmax=vmax,
            interpolation="nearest",
        )
        ax.set_title(name)
        plt.colorbar(im, ax=ax, fraction=0.046)
    fig.suptitle(title, fontsize=16)
    fig.tight_layout()
    plt.show()
````
